```python
import json, os
from tqdm import tqdm
```

```python
new_version_folder = "/scratch/lamdo/IRB/intermediate_results"
old_version_folder = "/scratch/lamdo/IRB/step7"
```

```python
new_version_filenames = os.listdir(os.path.join(new_version_folder, "step0"))
new_version_files = [os.path.join(new_version_folder, file) for file in new_version_filenames]

old_version_filenames = os.listdir(old_version_folder)
old_version_files = [os.path.join(old_version_folder, file) for file in old_version_filenames]
```

```python
all_new_version_filenames = set([])
for i in range(len(new_version_files)):
    filename = new_version_filenames[i]
    all_new_version_filenames.add(filename.replace('.json', ""))
```

```python
old_data = []
for i in tqdm(range(len(old_version_filenames))):
    path = old_version_files[i]
    with open(path) as f:
        temp = json.load(f)
        for line in temp:
            if line.get("title") in all_new_version_filenames:
                old_data.append(line)
```

```python
len(old_data)
```

```python
all_facts_old_ver = set([])
for line in old_data:
    paragraphs = line["paragraphs"]
    for para in paragraphs:
        validated_urls = para["validated_urls"]
        for item in validated_urls:
            modified_fact = item["fact"]
            all_facts_old_ver.add(modified_fact)

old_data[0]["paragraphs"][0]["validated_urls"]
```
