```python
import json
```

```python
qrel_path = "/scratch/lamdo/IRB/intermediate_results/step5/qrels.json"
```

```python
with open(qrel_path) as f:
    qrels = json.load(f)
```

```python
rel = 0
count = 0
for query_id in qrels:
    for doc_id in qrels[query_id]:
        rel += qrels[query_id][doc_id]
        count += 1
```

```python
rel / count
```
