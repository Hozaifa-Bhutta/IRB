```python
import json, random, os
import pandas as pd
```

```python
data_path = "/scratch/lamdo/IRB/runs/7March2026/step6"

with open(os.path.join(data_path, "queries.jsonl")) as f:
    queries = [json.loads(line) for line in f]

with open(os.path.join(data_path, "answers.jsonl")) as f:
    answers = [json.loads(line) for line in f]
```

```python
queries = [item for item in queries if not item["_id"].startswith("~")]
answers = [item for item in answers if not item["_id"].startswith("~")]
```

```python
sampled_ids = random.sample(range(len(queries)), 50)

sampled_queries = [queries[i] for i in sampled_ids]
sampled_answers = [answers[i] for i in sampled_ids]
```

```python
data = []
for q, a in zip(sampled_queries, sampled_answers):
    qtext = q["text"]
    ashort = a["short"]
    akps = "\n\n".join(a["text"])

    data.append({"query": qtext, "keypoints": akps, "answer": ashort})
```

```python
df = pd.DataFrame(data)
```

```python
df.to_csv("small_human_eval_7March2026.csv", index = False)
```
