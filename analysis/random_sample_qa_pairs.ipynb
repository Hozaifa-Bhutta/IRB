```python
import json, random, os
import pandas as pd
```

```python
data_path = "/home/lamdo/IllinoisRetrievalBenchmark/test_run_gitig_7/step5"

num_samples_each_type = 7

data_path_1_citations = os.path.join(data_path, "1_citations")
data_path_2_citations = os.path.join(data_path, "2_citations")
data_path_3_citations = os.path.join(data_path, "3_citations")


def list_of_text_to_bullet_points(list_of_text):
    return "\n".join(['+ '+ item for item in list_of_text])

sampled_queries = []
sampled_answers = []
for num_citations in range(1, 4):
    path = os.path.join(data_path, f"{num_citations}_citations")

    queries_path = os.path.join(path, "queries.jsonl")
    answers_path = os.path.join(path, "answers.jsonl")

    queries = []
    with open(queries_path) as f:
        for line in f:
            jline = json.loads(line)
            queries.append(jline)

    answers = []
    with open(answers_path) as f:
        for line in f:
            jline = json.loads(line)
            answers.append(jline)
    
    assert len(queries) == len(answers)

    sampled_indices = random.sample(range(len(queries)), num_samples_each_type)

    sampled_queries.extend([queries[i]["text"] for i in sampled_indices])
    sampled_answers.extend([list_of_text_to_bullet_points(answers[i]["text"]) for i in sampled_indices])

```

```python
df = pd.DataFrame()
df["queries"] = sampled_queries
df["answers"] = sampled_answers
```
