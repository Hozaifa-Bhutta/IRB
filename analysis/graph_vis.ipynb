```python
import networkx as nx
import matplotlib.pyplot as plt

# Your Knowledge Graph data
kg_data = [
            {
                "head": "Joseph Paul O'Brien-Whitmarsh",
                "head_type": "Person",
                "relation": "is",
                "tail": "Irish professional footballer",
                "tail_type": "Occupation",
                "head_coverage": [
                    0
                ],
                "tail_coverage": [
                    0
                ]
            },
            {
                "head": "Joseph Paul O'Brien-Whitmarsh",
                "head_type": "Person",
                "relation": "plays as",
                "tail": "midfielder",
                "tail_type": "Position",
                "head_coverage": [
                    0
                ],
                "tail_coverage": [
                    0
                ]
            },
            {
                "head": "Joseph Paul O'Brien-Whitmarsh",
                "head_type": "Person",
                "relation": "plays for",
                "tail": "Southampton",
                "tail_type": "Soccer team",
                "head_coverage": [
                    0
                ],
                "tail_coverage": [
                    0
                ]
            },
            {
                "head": "Joseph Paul O'Brien-Whitmarsh",
                "head_type": "Person",
                "relation": "joined on loan",
                "tail": "Accrington Stanley",
                "tail_type": "Soccer team",
                "head_coverage": [
                    0
                ],
                "tail_coverage": [
                    0
                ]
            },
            {
                "head": "Joseph Paul O'Brien-Whitmarsh",
                "head_type": "Person",
                "relation": "loan duration",
                "tail": "remainder of the 2024-25 season",
                "tail_type": "Time period",
                "head_coverage": [
                    0
                ],
                "tail_coverage": [
                    0
                ]
            }
        ]


# 1. Initialize a Directed Graph
G = nx.DiGraph()

# 2. Add nodes and edges (head -> tail) with the relation as an edge attribute
for item in kg_data:
    G.add_edge(item["head"], item["tail"], label=item["relation"])

# 3. Set up the visual layout
plt.figure(figsize=(14, 8))
# spring_layout space nodes out nicely; 'k' adjusts distance between them
pos = nx.spring_layout(G, k=0.8, seed=42) 

# 4. Draw the nodes and text
nx.draw(G, pos, 
        with_labels=True, 
        node_color='skyblue', 
        node_size=3500, 
        font_size=10, 
        font_weight='bold', 
        edge_color='gray', 
        arrows=True,
        arrowsize=20)

# 5. Draw the relation labels on the edges
edge_labels = nx.get_edge_attributes(G, 'label')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9, font_color='red')

# 6. Display the graph
# plt.title("Tokio Express Lego Spill - Knowledge Graph", fontsize=14)
plt.axis('off') # Hide the axes
plt.tight_layout()
plt.show()
```
