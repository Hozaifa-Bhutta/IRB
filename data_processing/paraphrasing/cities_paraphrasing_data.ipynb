```python
import json
```

```python
with open("./data/states.json") as f:
    data = json.load(f)
```

```python
res = {}
for line in data:
    to_append = {
        "name": line["name"],
        "type": line["type"],
        "lat": float(line["latitude"]),
        "long": float(line["longitude"]),
    }

    res[line["name"]] = to_append
```

```python
with open("test.json", "w") as f:
    json.dump(res, f, indent = 4)
```
