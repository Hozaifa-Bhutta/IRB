```python
import json
import pandas as pd
from tqdm import tqdm
```

```python
df = pd.read_csv("./data/JEOPARDY_CSV.csv")
```

```python
df
```

```python
test = []
for line in tqdm(df.to_dict("records")):
    if not (line.get(" Question") and line.get(" Answer")): continue
    if line[" Value"] != "$2000": continue
    answer = line[" Answer"]
    if str(answer).isnumeric():
        test.append([line[" Category"], line[" Question"], line[" Answer"]])
```

```python
paraphrase = {}

for category, question, answer in test:
    try:
        answer = float(answer)
        if int(answer) == answer:
            answer = int(answer)
        
    except Exception: 
        print("no")
        continue

    if any([undesired in question for undesired in ["href", "http"]]): continue

    if answer not in paraphrase: paraphrase[answer] = []
    paraphrase[answer].append(f"{category}. {question}")
```

```python
with open("_test.json", "w") as f:
    json.dump(paraphrase, f, indent = 4)
```
