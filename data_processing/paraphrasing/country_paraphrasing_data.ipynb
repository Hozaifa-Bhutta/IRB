```python
import pandas as pd
```

```python
df = pd.read_csv("./data/countries.csv")
```

```python
countries_data = df.to_dict(orient='records')
```

```python
countries_data[100:110]
```

```python
def neighbor_trivia(data):
    code2name = {}

    for line in data:
        code = line.get("cca3")
        name = line.get("name.common")
        if code and name:
            code2name[code] = name


    res = {}

    for line in data:
        borders_str = line.get("borders")
        all_names = [line.get("name.common"), line.get("name.official")]
        if not isinstance(borders_str, str): continue

        borders = [item.strip() for item in borders_str.split(",")]
        borders = [code2name.get(item) for item in borders]
        if not any([item is None for item in borders]):
            for name in all_names:
                res[name] = f"the country that borders {', '.join(borders)}"

    return res


def area_trivia(data):
    res = {}

    for line in data:
        all_names = [line.get("name.common"), line.get("name.official")]
        area = line.get("area")
        if area:
            for name in all_names:
                res[name] = f"the country whose total area is {area} square kilometers"

    return res

def calling_code_trivia(data):
    res = {}

    for line in data:
        all_names = [line.get("name.common"), line.get("name.official")]
        calling_code = line.get("callingCodes")

        if isinstance(calling_code, str) and not "," in calling_code:
            for name in all_names:
                res[name] = f"the country whose calling code is {calling_code}"

    return res
```

```python
area_trivia_data = area_trivia(countries_data)
neighbor_trivia_data = neighbor_trivia(countries_data)
calling_code_trivia_data = calling_code_trivia(countries_data)
```

```python
all_countries = set(area_trivia_data.keys()).union(set(neighbor_trivia_data.keys())).union(set(calling_code_trivia_data.keys()))
trivia_data = {}
for country in all_countries:
    at = area_trivia_data.get(country)
    nt = neighbor_trivia_data.get(country)
    cct = calling_code_trivia_data.get(country)

    to_update = [at, nt, cct]
    to_update = [item for item in to_update if item]

    trivia_data[country] = to_update
```

```python
import json
print(json.dumps(trivia_data, indent = 4))
```
