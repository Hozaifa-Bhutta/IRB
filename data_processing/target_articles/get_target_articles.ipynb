```python
import mwparserfromhell

def get_mentioned_pages(wikitext):
    """
    Parses wikitext and returns a list of mentioned Wikipedia page titles.
    """
    # 1. Parse the raw wikitext into a Wikicode object
    wikicode = mwparserfromhell.parse(wikitext)
    
    # 2. Extract all wikilink nodes (e.g., [[Page Title]] or [[Page Title|Display Text]])
    wikilinks = wikicode.filter_wikilinks()
    
    page_titles = []
    
    for link in wikilinks:
        # 3. Extract just the title portion of the link
        # .strip_code() removes any stray wikitext formatting (like italics) inside the title
        title = str(link.title.strip_code()).strip()
        
        # 4. Optional: Filter out non-article namespaces (Categories, Files, etc.)
        # Wikipedia uses colons for namespaces. If you only want standard articles, 
        # you can exclude titles containing standard namespace prefixes.
        namespaces_to_ignore = ("Category:", "File:", "Image:", "Wikipedia:", "Help:", "Template:")
        if not title.startswith(namespaces_to_ignore):
            page_titles.append(title)
            
    return page_titles

# --- Example Usage ---
mentioned_pages = set(["2025", "2026"])
for year in [2025, 2026]:
    with open(f"/scratch/lamdo/IRB/current_events/{year}.txt") as f:
        sample_wikitext = f.read()

    mentioned_pages.update(get_mentioned_pages(sample_wikitext))
```

```python
with open("target_articles_2025_2026.txt", "w") as f:
    f.write("\n".join(mentioned_pages))
```
